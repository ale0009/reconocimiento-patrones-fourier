# Demostración de Fourier-Mellin

Genera una letra transformada y visualiza la imagen, su espectro, el remuestreo log-polar y las firmas marginales.

```python
import numpy as np
import matplotlib.pyplot as plt
from fourier_patrones.datos import imagen_base, transformar
from fourier_patrones.transformadas import espectro_magnitud, remuestrear_log_polar, fourier_mellin_analitica
from fourier_patrones.firmas import firma_desde_matriz

imagen = transformar(imagen_base('A'), np.random.default_rng(12))
espectro = espectro_magnitud(imagen)
log_polar = remuestrear_log_polar(espectro, sigma=0.5)
fm = fourier_mellin_analitica(imagen, sigma=0.5)
firma = firma_desde_matriz(fm)
n = fm.shape[0]
fig, ejes = plt.subplots(1, 4, figsize=(13, 3))
ejes[0].imshow(imagen, cmap='gray'); ejes[0].set_title('Imagen')
ejes[1].imshow(np.log1p(espectro), cmap='magma'); ejes[1].set_title('Magnitud FFT 2D')
ejes[2].imshow(log_polar, cmap='magma', aspect='auto'); ejes[2].set_title('Log-polar')
ejes[3].plot(firma[:n], label='Filas'); ejes[3].plot(firma[n:], label='Columnas'); ejes[3].set_title('Firmas 1D'); ejes[3].legend()
for eje in ejes: eje.axis('off') if eje is not ejes[3] else None
plt.tight_layout()
```

La firma concatena las sumas por filas y columnas de la magnitud de la segunda FFT; ambas partes se normalizan por norma L2.
